# NanoGPT Playground — Free GPU Training

One-click reproduction of the `--big` model (2 layers / 4 heads / 128-dim) on a **free Colab GPU/CPU**.
No setup beyond pressing **Runtime → Run all**.

What this does:
1. Clones `Flynntaggart26/nanogpt-playground`
2. Smoke-tests the NumPy transformer (~1 min)
3. Trains the big model on full tiny-Shakespeare
4. Plots the loss curve + prints samples
5. Gives you a `weights.json` to download into the web playground

```
!git clone -q https://github.com/Flynntaggart26/nanogpt-playground.git
%cd nanogpt-playground
!pip install -q numpy matplotlib
```

```
# Smoke test: 50 iters on the tiny-default model — proves backprop works here
!python train.py --iters 50 --batch 8
```

```
# Full training: big model (2L/4H/128-dim) on 1MB Shakespeare.
# ~15-30 min on Colab CPU. Feel free to lower --iters for a quick look.
!python train.py --big --full
```

```
import json
import matplotlib.pyplot as plt

w = json.load(open('weights.json'))
print('config:', w['cfg'])
print('final loss:', round(w['loss'], 3))
plt.figure()
plt.plot(w['loss_curve'])
plt.xlabel('checkpoint'); plt.ylabel('train loss'); plt.title('NanoGPT training curve')
plt.show()

for t in ('0.7', '1.0'):
    print(f'--- sample temp={t} ---')
    print(open(f'sample_t{t}.txt').read()[:400])
    print()
```
